# PoRep Market — Happy Path E2E Tests

Interactive notebook for exercising the full deal lifecycle against a live devnet or calibnet deployment.

## Deal lifecycle covered

```
proposeDeal  →  (auto-)acceptDeal  →  ValidatorFactory.create
  →  Validator.createRail  →  Client.transfer (completes deal)
  →  payment settlement (FilecoinPay)  →  terminateDeal
```

### Prerequisites
```
pip install web3 python-dotenv
```
Set the following env vars (or fill `CONFIG` directly below):
- `RPC_URL` – e.g. `http://127.0.0.1:1234/rpc/v1` (devnet) or Glif public endpoint
- `PRIVATE_KEY` – client wallet private key (hex, 0x-prefixed)
- `PROVIDER_PRIVATE_KEY` – SP owner wallet private key (only needed if auto-approve is off)
- `POREP_SERVICE_PRIVATE_KEY` – PoRep service bot key (for payment settlement steps)
- `USDFC_TOKEN` – USDFC ERC-20 token address on the target network

## 0. Setup

````python
import json
import os
import time
from pathlib import Path

from dotenv import load_dotenv
from web3 import Web3
from web3.middleware import ExtraDataToPOAMiddleware

load_dotenv()

# ── Configuration ─────────────────────────────────────────────────────────────
CONFIG = {
    # RPC
    "rpc_url": os.getenv("RPC_URL", "http://127.0.0.1:1234/rpc/v1"),

    # Wallets  (override via env vars or set directly)
    "client_key":         os.getenv("PRIVATE_KEY", ""),
    "provider_key":       os.getenv("PROVIDER_PRIVATE_KEY", ""),
    "porep_service_key":  os.getenv("POREP_SERVICE_PRIVATE_KEY", ""),

    # Deployed contract addresses  (devnet latest.json defaults)
    "porep_market":       "0xD20Fc289be410d7CBF3fFF2e3d83778ea21c3b7B",
    "sp_registry":        "0xc66fc2b7FD6aa03047e43a6CA2E23C2296CCddAA",
    "validator_factory":  "0x6CCCE1D85e75DC7B402e70f5D2aB6544D792dCb5",
    "sli_oracle":         "0x6948607aDbB10AA51D298E7759Df35564145da32",
    "sli_scorer":         "0xb5A34880C80A7bFb4159ab0E9Ce0dbEbbB2C3807",
    "client_sc":          "0x7B4EC6E508a319493D44eEaa92e77BF39ae43E34",
    "filecoin_pay":       "0x4A8F3C2e1D9B7f6A0E5c8b3D2a1F9e7C4B6d8a2E",

    # Token used for payment rails (USDFC)
    "usdfc_token":        os.getenv("USDFC_TOKEN", "0x0000000000000000000000000000000000000000"),

    # Deal parameters
    "provider_actor_id":  1000,     # Filecoin actor ID of the SP
    "deal_size_bytes":    34359738368,  # 32 GiB
    "price_per_sector_per_month": 1_000_000,  # in USDFC (wei-equivalent)
    "duration_days":      30,       # must be multiple of 30
    "manifest_location":  "ipfs://bafybeigdyrzt5sfp7udm7hu76uh7y26nf3efuylqabf3oclgtqy55fbzdi",

    # SLI requirements (0 = don't care)
    "retrievability_bps": 9500,  # 95%
    "bandwidth_mbps":     100,
    "latency_ms":         0,     # don't care
    "indexing_pct":       0,     # don't care
}

# ── ABI paths ─────────────────────────────────────────────────────────────────
ABI_DIR = Path("../abis")

def load_abi(name: str):
    path = ABI_DIR / f"{name}.json"
    with open(path) as f:
        return json.load(f)

print("Configuration loaded.")
````

````python
# ── Web3 connection ───────────────────────────────────────────────────────────
w3 = Web3(Web3.HTTPProvider(CONFIG["rpc_url"]))
w3.middleware_onion.inject(ExtraDataToPOAMiddleware, layer=0)  # needed for Filecoin

assert w3.is_connected(), f"Cannot connect to {CONFIG['rpc_url']}"
chain_id = w3.eth.chain_id
print(f"Connected  chain_id={chain_id}  latest_block={w3.eth.block_number}")

# ── Accounts ──────────────────────────────────────────────────────────────────
client_acct   = w3.eth.account.from_key(CONFIG["client_key"])   if CONFIG["client_key"]         else None
provider_acct = w3.eth.account.from_key(CONFIG["provider_key"]) if CONFIG["provider_key"]        else None
service_acct  = w3.eth.account.from_key(CONFIG["porep_service_key"]) if CONFIG["porep_service_key"] else None

print(f"Client   : {client_acct.address   if client_acct   else 'NOT SET'}")
print(f"Provider : {provider_acct.address  if provider_acct  else 'NOT SET'}")
print(f"PoRepSvc : {service_acct.address   if service_acct   else 'NOT SET'}")

# ── Contract instances ────────────────────────────────────────────────────────
porep_market     = w3.eth.contract(address=CONFIG["porep_market"],      abi=load_abi("PoRepMarket"))
sp_registry      = w3.eth.contract(address=CONFIG["sp_registry"],       abi=load_abi("SPRegistry"))
validator_factory = w3.eth.contract(address=CONFIG["validator_factory"], abi=load_abi("ValidatorFactory"))
sli_oracle       = w3.eth.contract(address=CONFIG["sli_oracle"],         abi=load_abi("SLIOracle"))

print("\nContracts loaded.")
````

````python
# ── Transaction helpers ───────────────────────────────────────────────────────
def send_tx(fn, account, value=0):
    """Build, sign, send a transaction and wait for receipt."""
    nonce = w3.eth.get_transaction_count(account.address)
    gas_price = w3.eth.gas_price
    tx = fn.build_transaction({
        "from": account.address,
        "nonce": nonce,
        "gasPrice": gas_price,
        "value": value,
        "chainId": chain_id,
    })
    signed = account.sign_transaction(tx)
    tx_hash = w3.eth.send_raw_transaction(signed.raw_transaction)
    receipt = w3.eth.wait_for_transaction_receipt(tx_hash, timeout=120)
    status = "✅ success" if receipt["status"] == 1 else "❌ reverted"
    print(f"{status}  tx={tx_hash.hex()}  gas_used={receipt['gasUsed']}")
    return receipt

def decode_event(contract, receipt, event_name):
    """Extract the first matching event from a receipt."""
    event = getattr(contract.events, event_name)()
    logs = event.process_receipt(receipt)
    return logs[0]["args"] if logs else None

# ── State helpers ─────────────────────────────────────────────────────────────
DEAL_STATES = {0: "Proposed", 1: "Accepted", 2: "Completed", 3: "Rejected", 4: "Terminated"}

def fmt_deal(proposal):
    state_str = DEAL_STATES.get(proposal["state"], str(proposal["state"]))
    print(f"  dealId           : {proposal['dealId']}")
    print(f"  client           : {proposal['client']}")
    print(f"  provider (actor) : {proposal['provider']}")
    print(f"  state            : {state_str}")
    print(f"  validator        : {proposal['validator']}")
    print(f"  railId           : {proposal['railId']}")
    print(f"  manifestLocation : {proposal['manifestLocation']}")
    t = proposal["terms"]
    print(f"  terms.sizeBytes  : {t['dealSizeBytes']:,}")
    print(f"  terms.price/mo   : {t['pricePerSectorPerMonth']}")
    print(f"  terms.duration   : {t['durationDays']} days")
    r = proposal["requirements"]
    print(f"  req.retrievability : {r['retrievabilityBps']} bps")
    print(f"  req.bandwidth      : {r['bandwidthMbps']} Mbps")
    print(f"  req.latency        : {r['latencyMs']} ms")
    print(f"  req.indexing       : {r['indexingPct']}%")

print("Helpers ready.")
````

## 1. Preflight checks
Verify connectivity, balances, and that at least one active provider exists in the registry.

````python
# Chain / block
block = w3.eth.get_block("latest")
print(f"Latest block : {block['number']}  (timestamp {block['timestamp']})")

# Client balance
if client_acct:
    bal = w3.eth.get_balance(client_acct.address)
    print(f"Client balance : {w3.from_wei(bal, 'ether')} FIL")
    assert bal > 0, "Client wallet has no FIL – fund it before running txs"

# PoRepMarket sanity
max_duration = porep_market.functions.MAX_DEAL_DURATION_DAYS().call()
print(f"MAX_DEAL_DURATION_DAYS : {max_duration}")

print("\nPreflight OK ✅")
````

---
## 2. SP Registry — inspect providers

````python
providers = sp_registry.functions.getProviders().call()
print(f"Registered providers ({len(providers)}): {providers}\n")

for actor_id in providers:
    info = sp_registry.functions.getProviderInfo(actor_id).call()
    # info is a tuple; field order matches ProviderInfo struct
    org, payee, paused, blocked, capabilities, avail, committed, pending, price = info
    status = "BLOCKED" if blocked else ("PAUSED" if paused else "active")
    print(f"  Provider {actor_id} [{status}]")
    print(f"    org       : {org}")
    print(f"    payee     : {payee}")
    print(f"    available : {avail:,} bytes")
    print(f"    committed : {committed:,} bytes")
    print(f"    pending   : {pending:,} bytes")
    print(f"    price/mo  : {price}")
    print(f"    sli caps  : retrievability={capabilities[0]} bps  "
          f"bandwidth={capabilities[1]} Mbps  "
          f"latency={capabilities[2]} ms  "
          f"indexing={capabilities[3]}%")
    print()
````

````python
# Check target provider SLI attestation
actor_id = CONFIG["provider_actor_id"]
attestation = sli_oracle.functions.getAttestation(actor_id).call()
last_update, slis = attestation
print(f"SLI attestation for actor {actor_id}:")
print(f"  lastUpdate (block) : {last_update}")
print(f"  retrievability     : {slis[0]} bps")
print(f"  bandwidth          : {slis[1]} Mbps")
print(f"  latency            : {slis[2]} ms")
print(f"  indexing           : {slis[3]}%")
````

---
## 3. Propose a deal

`PoRepMarket.proposeDeal(requirements, terms, manifestLocation)` — called by the client wallet.

The registry automatically selects the least-committed matching provider.
If the provider's `pricePerSectorPerMonth` ≤ deal price → the deal **auto-accepts**.

````python
assert client_acct, "Set PRIVATE_KEY in your environment"

requirements = (
    CONFIG["retrievability_bps"],
    CONFIG["bandwidth_mbps"],
    CONFIG["latency_ms"],
    CONFIG["indexing_pct"],
)
terms = (
    CONFIG["deal_size_bytes"],
    CONFIG["price_per_sector_per_month"],
    CONFIG["duration_days"],
)

fn = porep_market.functions.proposeDeal(requirements, terms, CONFIG["manifest_location"])
receipt = send_tx(fn, client_acct)

# Extract dealId from DealProposalCreated event
ev = decode_event(porep_market, receipt, "DealProposalCreated")
assert ev, "DealProposalCreated event not found in receipt"

DEAL_ID = ev["dealId"]
print(f"\nDeal proposed  dealId={DEAL_ID}  provider={ev['provider']}")
````

````python
# Read back the proposal
proposal = porep_market.functions.getDealProposal(DEAL_ID).call()
print("Deal proposal:")
fmt_deal(proposal)

AUTO_ACCEPTED = proposal["state"] == 1  # Accepted
print(f"\nAuto-accepted: {AUTO_ACCEPTED}")
````

---
## 4. Accept the deal  _(skip if auto-accepted)_

The controlling address of the SP actor calls `acceptDeal(dealId)`.

````python
if AUTO_ACCEPTED:
    print("Deal was auto-accepted — skipping manual acceptDeal step.")
else:
    assert provider_acct, "Set PROVIDER_PRIVATE_KEY to accept the deal manually"
    fn = porep_market.functions.acceptDeal(DEAL_ID)
    receipt = send_tx(fn, provider_acct)
    ev = decode_event(porep_market, receipt, "DealAccepted")
    print(f"Deal accepted  dealId={ev['dealId']}  provider={ev['provider']}")

# Verify state == Accepted
proposal = porep_market.functions.getDealProposal(DEAL_ID).call()
assert proposal["state"] == 1, f"Expected state=Accepted, got {DEAL_STATES[proposal['state']]}"
print("\nState: Accepted ✅")
````

---
## 5. Create a Validator instance

`ValidatorFactory.create(dealId)` deploys a per-deal `BeaconProxy` for the `Validator` contract.  
The new Validator automatically calls `PoRepMarket.updateValidator(dealId)` in its initializer,
so no extra transaction is needed.

````python
fn = validator_factory.functions.create(DEAL_ID)
receipt = send_tx(fn, client_acct)

VALIDATOR_ADDR = validator_factory.functions.getInstance(DEAL_ID).call()
print(f"Validator deployed at {VALIDATOR_ADDR}")

# Verify the deal now points to the validator
proposal = porep_market.functions.getDealProposal(DEAL_ID).call()
assert proposal["validator"].lower() == VALIDATOR_ADDR.lower(), "Validator address mismatch on deal"
print("PoRepMarket.validator field updated ✅")

validator = w3.eth.contract(address=VALIDATOR_ADDR, abi=load_abi("Validator"))
````

---
## 6. Create payment rail

`Validator.createRail(token)` — called by the **client** wallet.  
This creates a FilecoinPay rail and calls `PoRepMarket.updateRailId(dealId, railId)` internally.

````python
usdfc = Web3.to_checksum_address(CONFIG["usdfc_token"])
fn = validator.functions.createRail(usdfc)
receipt = send_tx(fn, client_acct)

# Rail ID is now stored on the deal
proposal = porep_market.functions.getDealProposal(DEAL_ID).call()
RAIL_ID = proposal["railId"]
assert RAIL_ID != 0, "railId was not set after createRail"
print(f"Rail created  railId={RAIL_ID}")

min_epochs = validator.functions.getMinEpochsBetweenSettlements().call()
print(f"Min epochs between settlements : {min_epochs}  (~{min_epochs / 2880:.1f} days)")
````

## 8. Inspect deal state

````python
proposal = porep_market.functions.getDealProposal(DEAL_ID).call()
print("Current deal state:")
fmt_deal(proposal)
````

````python
# List all deals in Completed state (ready for payment settlement)
completed = porep_market.functions.getCompletedDeals().call()
print(f"Completed deals ready for settlement: {len(completed)}")
for d in completed:
    print(f"  dealId={d['dealId']}  provider={d['provider']}  railId={d['railId']}")
````

````python
# SP Registry — check capacity committed vs pending after deal progress
actor_id = CONFIG["provider_actor_id"]
info = sp_registry.functions.getProviderInfo(actor_id).call()
org, payee, paused, blocked, capabilities, avail, committed, pending, price = info
print(f"Provider {actor_id} capacity:")
print(f"  available : {avail:,} bytes")
print(f"  committed : {committed:,} bytes")
print(f"  pending   : {pending:,} bytes")
````

---
## 9. Payment settlement  _(PoRep service bot)_

Settlement is triggered by the PoRep bot calling `FilecoinPay.settleRail(railId)`.  
The FilecoinPay contract then calls back into `Validator.validatePayment(...)`.  
Below we call the Validator's `modifyRailPayment` and `disableFutureRailPayments`
as the PoRep service to show the bot-side interaction.

````python
# Requires POREP_SERVICE_PRIVATE_KEY and deal to be Completed

if service_acct and proposal["state"] == 2:  # Completed
    # Example: update payment rate via the service bot
    fn = validator.functions.modifyRailPayment(RAIL_ID)
    receipt = send_tx(fn, service_acct)
    print("Rail payment rate updated ✅")
else:
    if not service_acct:
        print("Skipping — set POREP_SERVICE_PRIVATE_KEY to exercise settlement steps.")
    else:
        print(f"Skipping — deal is not Completed yet (state={DEAL_STATES[proposal['state']]}).")
````

---
## 10. Terminate deal  _(via Validator)_

The Validator calls `PoRepMarket.terminateDeal(dealId, terminator, endEpoch)`.  
This moves the deal to `Terminated`, releases provider capacity, and
removes the deal from the settlement queue.

````python
# terminateDeal must be called BY the registered Validator contract, not directly.
# The PoRep service bot triggers it via Validator.terminateRail(railId).

if service_acct and proposal["state"] == 2:  # Completed
    end_epoch = w3.eth.block_number + 1  # terminate at next block for testing
    fn = validator.functions.terminateRail(RAIL_ID)
    receipt = send_tx(fn, service_acct)

    proposal = porep_market.functions.getDealProposal(DEAL_ID).call()
    assert proposal["state"] == 4, "Expected state=Terminated"
    print(f"Deal terminated ✅  state={DEAL_STATES[proposal['state']]}")
else:
    print("Skipping termination — deal must be Completed and POREP_SERVICE_PRIVATE_KEY set.")
````

---
## 11. Final state summary

````python
proposal = porep_market.functions.getDealProposal(DEAL_ID).call()
print("═" * 50)
print("FINAL DEAL STATE")
print("═" * 50)
fmt_deal(proposal)

remaining_completed = porep_market.functions.getCompletedDeals().call()
print(f"\nDeals in settlement queue : {len(remaining_completed)}")

info = sp_registry.functions.getProviderInfo(CONFIG["provider_actor_id"]).call()
_, _, _, _, _, avail, committed, pending, _ = info
print(f"Provider capacity — avail={avail:,}  committed={committed:,}  pending={pending:,}")

print("\nHappy path run complete.")
````
